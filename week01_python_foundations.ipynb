```python
trips = 25
avg_fare = 220.0
fuel_cost = 1100.0
rider_name = "Kevin"
```

```python
gross_income = trips * avg_fare

net_income = gross_income - fuel_cost

margin_pct = (net_income / gross_income) * 100

print(gross_income)
print(net_income)
print(margin_pct)
```

```python
print(type(trips))
print(type(avg_fare))
print(type(fuel_cost))
print(type(rider_name))
```

```python
print(f"{rider_name} earned KSh {gross_income:,.2f} gross, KSh {net_income:,.2f} net, with a margin of {margin_pct:.1f}%.")
```
